El medio sumador tiene dos entradas (**A**, **B**) y dos salidas:

#### Tabla de verdad del medio sumador

| A | B | SUMA | ACARREO |

La actividad consiste en entrenar un **perceptrón simple** para cada una de las salidas y descubrir qué ocurre.
