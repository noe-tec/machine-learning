## The Problem with Pure Generation → The Solution: RAG

GPT is powerful but has two well-known weaknesses:

| Problem | Consequence |
|---------|------------|
| **Hallucination** | May invent fake citations, wrong statistics, or events |
| **Stale knowledge** | Knows nothing about events after training cutoff |

What if we could hand the model fresh, verified information at answer time?
That is exactly what **RAG** does.

---

---
## RAG: Retrieval-Augmented Generation

### What is RAG?

RAG stands for Retrieval-Augmented Generation.
It was introduced by Lewis et al. at Facebook AI Research in 2020.

### The Core Idea

Instead of asking the model to *remember* every fact, we:
1. Store knowledge in an external database (PDF, website, knowledge graph…)
2. Retrieve the most relevant passages when a question arrives
3. Generate an answer grounded in those real passages

```
┌─────────────────────────────────────────────────────────┐
│  WITHOUT RAG                                            │
│  Question ──────────────────────────────→ LLM → Answer  │
│                                     (from memory only)  │
└─────────────────────────────────────────────────────────┘

┌─────────────────────────────────────────────────────────┐
│  WITH RAG                                               │
│  Question ──→ Retriever ──→ Relevant Chunks ──→ LLM     │
│                  ↑                              ↓       │
│            Document store               Grounded Answer │
└─────────────────────────────────────────────────────────┘
```

### Why not send the whole document?

- Documents can be thousands of pages — too large for the model context window
- Retrieving only the relevant parts is faster and more focused
- The same system scales to millions of documents

### Split the Document into Chunks

A full document is often too long to give entirely to a language model,
and we also need searchable pieces. We split the text into smaller, overlapping segments.

#### Why overlap?

Imagine the document says:
*"…Einstein was born in Ulm. He later moved to Switzerland."*

If we split at the period, one chunk ends with *"…born in Ulm."* and the next begins
with *"He later moved…"* — we lost the subject!

With overlap, the last few characters of one chunk repeat at the start of the next,
preserving context across boundaries.

```
Full text:  AAAAAABBBBBBCCCCCCDDDDDDEEEEE

Chunk 1:    AAAAAABBBBB
Chunk 2:          BBBBBCCCCC    ← overlap with chunk 1
Chunk 3:                CCCCCDDDDD
...
```


````python
import re
# RAG: Split the document into overlapping chunks

def split_into_chunks(text, chunk_size=500, overlap=80):
    '''
    Split a text string into overlapping chunks.

    Args:
        text       : The full document text (one long string)
        chunk_size : Target number of characters per chunk
        overlap    : Characters shared between consecutive chunks

    Returns:
        List of text strings (the chunks)
    '''
    # 1. Normalise whitespace: collapse multiple spaces/newlines into a single space
    text = re.sub(r"\s+", " ", text).strip()

    chunks = []
    start  = 0

    while start < len(text):
        end = start + chunk_size

        # 2. Try to break at a sentence boundary (. ? !) to avoid cutting mid-sentence
        if end < len(text):
            last_break = max(
                text.rfind(". ", start, end),
                text.rfind("? ", start, end),
                text.rfind("! ", start, end),
            )
            if last_break > start + overlap:
                end = last_break + 2     # include the punctuation and trailing space

        chunk = text[start:end].strip()
        if chunk:                        # only keep non-empty chunks
            chunks.append(chunk)

        # 3. Move forward, keeping `overlap` characters of context
        start = end - overlap

    return chunks

# ─── Apply the chunking function ─────────────────────────────────────────────
CHUNK_SIZE = 500   # characters per chunk (try values between 300 and 800)
OVERLAP    = 80    # characters of overlap between consecutive chunks

chunks = split_into_chunks(document_text, chunk_size=CHUNK_SIZE, overlap=OVERLAP)

print("Step 2: Chunking")
print("=" * 65)
print(f"Chunk size (target) : {CHUNK_SIZE} characters")
print(f"Overlap             : {OVERLAP} characters")
print(f"Total chunks        : {len(chunks)}")
print()

# ─── Inspect the first, a middle, and the last chunk ─────────────────────────
sample_indices = [0, len(chunks) // 2, len(chunks) - 1]
labels         = ["First", "Middle", "Last"]

for idx, label in zip(sample_indices, labels):
    chunk = chunks[idx]
    print(f"--- Chunk {idx} ({label}) ---")
    print_wrapped(chunk,90)
    print(f"[{len(chunk)} characters]")
    print()
````

### Create Embeddings

A language model cannot directly compare two text strings for meaning.
We need to convert each chunk into a vector embeddings.

#### How do embeddings capture meaning?

The embedding model is trained so that semantically similar sentences produce similar vectors.

```
"The sky is blue."      → [ 0.12, -0.34,  0.78, …]  (384 numbers)
"The sky appears blue." → [ 0.11, -0.31,  0.76, …]  ← very close!
"I love pizza."         → [-0.53,  0.21, -0.14, …]  ← very different
```

We use `sentence-transformers/all-MiniLM-L6-v2`:

| Property | Value |
|----------|-------|
| Size | ~90 MB |
| Output dimensions | 384 numbers per sentence |
| Speed on CPU | Fast (~500 sentences/sec) |
| Quality | Excellent for semantic similarity |


````python
import torch

# Device detection (CPU / GPU)
DEVICE      = 0 if torch.cuda.is_available() else -1   # 0 = first GPU, -1 = CPU
DEVICE_NAME = "GPU (CUDA)" if DEVICE == 0 else "CPU "
DEVICE_ST   = "cuda"       if DEVICE == 0 else "cpu"  # SentenceTransformer uses strings
print(f"Running on: {DEVICE_NAME}")
````

````python
# Compute embeddings for every chunk
import sentence_transformers
print("sentence_transformers", sentence_transformers.__version__)
import numpy as np
from sentence_transformers import SentenceTransformer
from transformers import pipeline


print("Step 3: Creating Embeddings")
print("=" * 65)
print("Loading: sentence-transformers/all-MiniLM-L6-v2 (~90 MB)")
print("First run downloads the model; subsequent runs use the cache.")
print()

# Load the sentence embedding model
embedding_model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2", device=DEVICE_ST)

print("Embedding model loaded!")
print()
print(f"Encoding {len(chunks)} chunks in batches...")

# Encode all chunks at once — batch processing is much faster than one-by-one
chunk_embeddings = embedding_model.encode(
    chunks,
    show_progress_bar=True,    # display a progress bar during encoding
    convert_to_numpy=True,     # return a NumPy array for easy math later
)

print()
print("Embedding complete!")
print()
print(f"Embedding matrix shape : {chunk_embeddings.shape}")
print(f"  → {chunk_embeddings.shape[0]} chunks")
print(f"  → each chunk = a vector of {chunk_embeddings.shape[1]} numbers")
print()

# ─── Quick sanity check: neighbouring chunks should be close in embedding space
vec0 = chunk_embeddings[0]
vec1 = chunk_embeddings[1]
sim  = np.dot(vec0, vec1) / (np.linalg.norm(vec0) * np.linalg.norm(vec1))
print(f"Sanity check — similarity between chunk 0 and chunk 1: {sim:.4f}")
print("(Adjacent chunks usually score high due to overlap.)")
````

### Retrieve Relevant Chunks

When the user asks a question we:
1. Embed the question using the same model used for the chunks
2. Compute cosine similarity between the question vector and every chunk vector
3. Return the top-k most similar chunks

#### Cosine Similarity — Intuition

```
              question ↗
                        ↘ chunk A   (small angle → HIGH similarity ≈ 0.9)
          chunk B ↙               (large angle → LOW similarity ≈ 0.2)
```

**Formula:**
```
                   A · B
sim(A, B)  =  ─────────────────
               ‖A‖  ×  ‖B‖
```

- **1.0** → vectors point in the same direction (identical meaning)
- **0.0** → vectors are perpendicular (unrelated topics)
- **−1.0** → opposite directions (antonyms)


````python
# Retrieve the most relevant chunks

def cosine_similarity(vec_a, vec_b):
    '''
    Compute cosine similarity between two NumPy 1-D vectors.
    Returns a float in the range [-1, 1]. Higher means more similar.
    '''
    dot   = np.dot(vec_a, vec_b)
    norms = np.linalg.norm(vec_a) * np.linalg.norm(vec_b)
    if norms == 0:
        return 0.0              # guard against zero-norm vectors
    return dot / norms


def retrieve_top_k(query, chunks, chunk_embeddings, embed_model, top_k=3):
    '''
    Find the top-k document chunks most relevant to a query.

    Args:
        query            : User question (plain text string)
        chunks           : List of text chunk strings
        chunk_embeddings : NumPy array of shape (n_chunks, embed_dim)
        embed_model      : SentenceTransformer instance (same as used for chunks)
        top_k            : Number of chunks to return

    Returns:
        List of (chunk_text, similarity_score) tuples, sorted best-first.
    '''
    # 1. Encode the question with the SAME model used for the chunks
    q_emb = embed_model.encode(query, convert_to_numpy=True)

    # 2. Compute similarity between the question and every chunk
    scores = []
    for idx, c_emb in enumerate(chunk_embeddings):
        score = cosine_similarity(q_emb, c_emb)
        scores.append((idx, score))

    # 3. Sort by score, highest first
    scores.sort(key=lambda x: x[1], reverse=True)

    # 4. Return the top-k (chunk text, score) pairs
    return [(chunks[idx], score) for idx, score in scores[:top_k]]


# ─── Test retrieval with a sample question ────────────────────────────────────
sample_query = "What does BERT stand for in this document?"

print("Step 4: Semantic Retrieval")
print("=" * 65)
print(f"Query: '{sample_query}'")
print()

retrieved = retrieve_top_k(
    query=sample_query,
    chunks=chunks,
    chunk_embeddings=chunk_embeddings,
    embed_model=embedding_model,
    top_k=3,
)

print(f"Top-3 most relevant chunks:")
print()
for rank, (chunk, score) in enumerate(retrieved, start=1):
    preview = chunk[:200] + ("..." if len(chunk) > 200 else "")
    print(f"Rank {rank}  |  similarity: {score:.4f}")
    print(f"  {preview}")
    print()
````

### Generate a Grounded Answer

We pass the retrieved chunks + the original question to a language model.
The model reads the context and generates an answer **grounded in the real document**.


#### The Prompt Template

```
Answer the question based only on the context below.

Context:
<retrieved chunk 1>
<retrieved chunk 2>
<retrieved chunk 3>

Question: <user's question>

Answer:
```

Because the answer is grounded in real retrieved text, the model is far less likely to hallucinate.


````python
# Load generation model and generate grounded answers

print("Loading Generation Model")

rag_gen = pipeline(
    task="text-generation",
    model="Qwen/Qwen2.5-0.5B-Instruct", # another option: destilgpt2
    max_new_tokens=200,
    device=DEVICE,
)

print("Generation model loaded!")
print()
````

````python
def build_prompt(query, retrieved_chunks, max_context_chars=1400):
    context_parts = [chunk for chunk, _score in retrieved_chunks]
    context = "\n\n".join(context_parts)

    if len(context) > max_context_chars:
        context = context[:max_context_chars] + "..."

    prompt = (
        "Answer the question using only the context below.\n"
        "Write a concise answer in 2 or 3 sentences.\n"
        "Do not repeat ideas.\n\n"
        f"Context:\n{context}\n\n"
        f"Question: {query}\n\n"
        "Answer:"
    )

    return prompt

def generate_answer(query, retrieved_chunks, rag_gen):
    prompt = build_prompt(query, retrieved_chunks)

    output = rag_gen(
        prompt,
        max_new_tokens=120,
        do_sample=False,
        pad_token_id=rag_gen.tokenizer.eos_token_id,
    )

    generated = output[0]["generated_text"]

    # Quitar el prompt si el modelo lo repite
    answer = generated[len(prompt):].strip()

    return answer


answer = generate_answer(sample_query, retrieved, rag_gen)

print(f"Question : {sample_query}")
print()
print("Retrieved context (top-3 chunks):")
for i, (chunk, score) in enumerate(retrieved, 1):
    print(f"  Chunk {i} (score {score:.4f}): {chunk[:80]}...")
print()
print("Generated Answer:")
print("-" * 65)
print_wrapped(answer, width=90)
print("-" * 65)
````

### Putting It All Together — The Complete RAG Pipeline

All five steps are now packaged into a single `rag()` function.
We test it with several different questions about the loaded document.


````python
# Complete RAG pipeline

def rag(query, chunks, chunk_embeddings, embed_model, generator,
        top_k=3, show_retrieved=True):
    '''
    End-to-end RAG: retrieve the most relevant chunks and generate a grounded answer.

    Args:
        query            : User question (string)
        chunks           : List of document chunk strings
        chunk_embeddings : Pre-computed embeddings for all chunks (NumPy array)
        embed_model      : SentenceTransformer for encoding queries
        generator        : Flan-T5 text-generation pipeline
        top_k            : Number of chunks to retrieve
        show_retrieved   : If True, print the retrieved chunks for inspection

    Returns:
        dict with keys: query, retrieved, answer
    '''
    print()
    print("╔" + "=" * 62 + "╗")
    print(f"  QUESTION: {query}")
    print("╚" + "=" * 62 + "╝")

    # Step A: Retrieve relevant chunks
    retrieved = retrieve_top_k(
        query=query,
        chunks=chunks,
        chunk_embeddings=chunk_embeddings,
        embed_model=embed_model,
        top_k=top_k,
    )

    if show_retrieved:
        print()
        print(f"  Top-{top_k} retrieved chunks:")
        for i, (chunk, score) in enumerate(retrieved, 1):
            #snippet = chunk[:110].replace("\n", " ")
            print(f"    [{i}] score={score:.4f}")
            print_wrapped(chunk, width=90)
            print()

    # Step B: Generate grounded answer
    answer = generate_answer(query, retrieved, generator)

    print()
    print("  ANSWER: \n")
    print_wrapped(answer, width=90)
    print()

    return {"query": query, "retrieved": retrieved, "answer": answer}


# ─── Run multiple test questions ──────────────────────────────────────────────
questions = [
    "What does BERT stand for in this document?",
    "Who created BERT and when?",
    "How was BERT trained?",
    "What does GPT stand for in this document?",
    "What is Orchard-Augmented Generation?",
    "What is the Pineapple Contradiction Test?",
]

results = []
for q in questions:
    r = rag(
        query=q,
        chunks=chunks,
        chunk_embeddings=chunk_embeddings,
        embed_model=embedding_model,
        generator=rag_gen,
        top_k=3,
        show_retrieved=False,   # set to True to inspect retrieved passages
    )
    results.append(r)
````
