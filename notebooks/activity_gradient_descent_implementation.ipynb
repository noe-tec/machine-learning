```python
# Required functions

def draw_line(gradient, x_center, y_center, distance, axis):
    """
    Draws a tangent line that represents the gradient at the point (x_center, y_center).
    """
    x_vals = np.linspace(x_center - distance, x_center + distance, 50)
    y_vals = gradient * (x_vals - x_center) + y_center

    axis.scatter(x_center, y_center, color='b', s=50)
    axis.plot(x_vals, y_vals, linestyle='--', color='r', zorder=10, linewidth=1)

    x_offset = 30 if x_center == 200 else 10
    text = r"$\frac{\partial J}{\partial w}$ = %d" % gradient

    axis.annotate(text,
                  fontsize=14,
                  xy=(x_center, y_center),
                  xycoords='data',
                  xytext=(x_offset, 10),
                  textcoords='offset points',
                  arrowprops=dict(arrowstyle="->"),
                  ha='left',
                  va='top')

def plot_gradients(x_data, y_labels, cost_function, gradient_function):
    """
    Visualizes the relationship between the cost and the parameter w, and shows the gradient field
    in a vector plot.
    """
    figure, axes = plt.subplots(1, 2, figsize=(12, 4))

    # ============
    # First plot
    # ============
    fixed_b = 100
    w_values = np.linspace(0, 400, 50)
    costs = np.array([cost_function(x_data, y_labels, w, fixed_b) for w in w_values])

    axes[0].plot(w_values, costs, linewidth=1)
    axes[0].set_title("Cost vs w with gradient (b = 100)")
    axes[0].set_xlabel("w")
    axes[0].set_ylabel("Cost")

    # Add gradient lines for certain values of w
    for test_w in [50, 200, 250]:
        grad_w, _ = gradient_function(x_data, y_labels, test_w, fixed_b)
        current_cost = cost_function(x_data, y_labels, test_w, fixed_b)
        draw_line(grad_w, test_w, current_cost, 30, axes[0])

    # ===============
    # Second plot
    # ===============
    b_grid, w_grid = np.meshgrid(np.linspace(-200, 200, 10),
                                 np.linspace(-100, 600, 10))
    grad_w_matrix = np.zeros_like(w_grid)
    grad_b_matrix = np.zeros_like(b_grid)

    for i in range(w_grid.shape[0]):
        for j in range(w_grid.shape[1]):
            grad_w_matrix[i, j], grad_b_matrix[i, j] = gradient_function(
                x_data, y_labels, w_grid[i, j], b_grid[i, j]
            )

    color_magnitude = np.sqrt((grad_b_matrix / 2)**2 + (grad_w_matrix / 2)**2)

    axes[1].set_title("Gradient field (vector plot)")
    Q = axes[1].quiver(w_grid, b_grid, grad_w_matrix, grad_b_matrix,
                       color_magnitude, units='width')
    axes[1].quiverkey(Q, 0.9, 0.9, 2, r'$2 \frac{m}{s}$', labelpos='E', coordinates='figure')
    axes[1].set_xlabel("w")
    axes[1].set_ylabel("b")


def within_limits(a, b, xlim, ylim):
    xlow, xhigh = xlim
    ylow, yhigh = ylim
    ax, ay = a
    bx, by = b
    if (ax > xlow and ax < xhigh) and (bx > xlow and bx < xhigh) \
        and (ay > ylow and ay < yhigh) and (by > ylow and by < yhigh):
        return True
    return False

def plot_contour_with_descent(x, y, history, axis, 
                              compute_cost_function,
                              w_range=[-50, 400, 5], 
                              b_range=[-450, 450, 5],
                              contour_levels=[0.1, 100, 1000, 5000, 10000, 25000, 50000],
                              resolution=5, final_w=200, final_b=100, step=10):
    """
    Draws the contour plot of the cost function J(w, b) and the path followed by gradient descent.

    Parameters:
        x, y (ndarray): Data and labels.
        history (list): List containing the parameter history [w, b].
        axis (matplotlib axis): Axis where the plot will be drawn.
        w_range, b_range (list): Range for the values of w and b.
        contour_levels (list): Cost function levels for the contour lines.
        resolution (int): Minimum distance between arrows for the path.
        final_w, final_b (float): Last values reached by w and b.
        step (int): Interval between points for drawing the path arrows.
    """

    b_vals, w_vals = np.meshgrid(np.arange(*b_range), np.arange(*w_range))
    z = np.zeros_like(b_vals)

    for i in range(w_vals.shape[0]):
        for j in range(w_vals.shape[1]):
            z[i][j] = compute_cost_function(x, y, w_vals[i][j], b_vals[i][j])

    contours = axis.contour(w_vals, b_vals, z, contour_levels, linewidths=2)
    axis.clabel(contours, inline=True, fmt='%1.0f', fontsize=10)
    axis.set_xlabel("w")
    axis.set_ylabel("b")
    axis.set_title("Contour plot of J(w, b) with the gradient descent path")

    # Draw path
    previous_point = history[0]
    for point in history[0::step]:
        distance = np.sqrt((previous_point[0] - point[0])**2 + (previous_point[1] - point[1])**2)
        if distance > resolution or point == history[-1]:
            if within_limits(point, previous_point, axis.get_xlim(), axis.get_ylim()):
                plt.annotate('', xy=point, xytext=previous_point, xycoords='data',
                             arrowprops={'arrowstyle': '->', 'color': 'b', 'lw': 3},
                             va='center', ha='center')
            previous_point = point
    return
```

## Gradient Descent Review

Let us recall that our linear model is:

$h_{w,b}(x^{(i)}) = wx^{(i)} + b$

In linear regression, training examples are used to adjust the parameters $w$, $b$ by minimizing an error measure between our predictions $\hat{y}=h_{w,b}(x)$ and the observed actual values $y$. We call this measure the “cost.”

In this case, we are using the mean squared error as the cost function:

$J(w,b) = \frac{1}{N} \sum\limits_{i = 0}^{N-1} (h_{w,b}(x^{(i)}) - y^{(i)})^2$


The gradient descent algorithm is described as:

$$\begin{align*} \text{repeat}&\text{ until convergence:} \; \lbrace \newline
\;  w &= w -  \alpha \frac{\partial J(w,b)}{\partial w} \tag{3}  \; \newline 
 b &= b -  \alpha \frac{\partial J(w,b)}{\partial b}  \newline \rbrace
\end{align*}$$
where the parameters $w$, $b$ are updated simultaneously.  
Therefore, to implement gradient descent, we need the partial derivatives of the cost with respect to the parameters:
$$
\begin{align}
\frac{\partial J(w,b)}{\partial w}  &= \frac{2}{N} \sum\limits_{i = 0}^{N-1} (h_{w,b}(x^{(i)}) - y^{(i)})x^{(i)} \tag{4}\\
  \frac{\partial J(w,b)}{\partial b}  &= \frac{2}{N} \sum\limits_{i = 0}^{N-1} (h_{w,b}(x^{(i)}) - y^{(i)}) \tag{5}\\
\end{align}
$$

Here *simultaneously* means that the partial derivatives for all parameters must be calculated before updating any of the parameters.

###  Gradient Descent
The following cell defines the gradient descent function that will be used to adjust the values of $w$ and $b$.

```python
def gradient_descent(data, labels, initial_w, initial_b, learning_rate, iterations, cost_function, gradient_function):
    """
    Performs gradient descent to adjust the parameters w and b.
    Updates w and b for the specified number of iterations using a
    specific learning rate.

    Parameters:
      data (ndarray (m,))          : Input data, m examples
      labels (ndarray (m,))        : Target values
      initial_w, initial_b (float) : Initial values of the parameters
      learning_rate (float)        : Learning rate
      iterations (int)             : Number of iterations of the algorithm
      cost_function                : Function that computes the cost
      gradient_function            : Function that computes the gradients

    Returns:
      final_w (float)              : Final value of parameter w
      final_b (float)              : Final value of parameter b
      cost_history (list)          : List containing the cost value at each iteration
      parameter_history (list)     : List containing the values of [w, b] at each iteration
    """

    cost_history = []
    parameter_history = []

    current_w = initial_w
    current_b = initial_b

    for step in range(iterations):
        grad_w, grad_b = gradient_function(data, labels, current_w, current_b)

        # Update the parameters
        current_w = current_w - learning_rate * grad_w
        current_b = current_b - learning_rate * grad_b

        # Save history if it is not too large
        if step < 100000:
            cost = cost_function(data, labels, current_w, current_b)
            cost_history.append(cost)
            parameter_history.append([current_w, current_b])

        # Print the status every certain number of iterations
        if step % math.ceil(iterations / 10) == 0:
            print(f"Iteration {step:4}: Cost {cost_history[-1]:0.2e} ",
                  f"grad_w: {grad_w: 0.3e}, grad_b: {grad_b: 0.3e}  ",
                  f"w: {current_w: 0.3e}, b: {current_b: 0.5e}")

    return current_w, current_b, cost_history, parameter_history
```
