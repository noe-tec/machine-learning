Entrenar ANN con MNIST

```python
# ==============================================
# MNIST - Estrategias de ajuste de redes neuronales
# ==============================================

import tensorflow as tf
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Flatten, Dropout
from tensorflow.keras.utils import to_categorical
import matplotlib.pyplot as plt

# Función para graficar
def plot_history(history, title="Curvas de entrenamiento"):
    plt.figure(figsize=(12,4))
    # Pérdida
    plt.subplot(1,2,1)
    plt.plot(history.history['loss'], label='Entrenamiento')
    plt.plot(history.history['val_loss'], label='Validación')
    plt.title(title + " - Pérdida")
    plt.legend()
    # Exactitud
    plt.subplot(1,2,2)
    plt.plot(history.history['accuracy'], label='Entrenamiento')
    plt.plot(history.history['val_accuracy'], label='Validación')
    plt.title(title + " - Exactitud")
    plt.legend()
    plt.show()

# 1. Cargar datos
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# 🔴 Caso 1: SIN normalización
# (más adelante haremos: x_train = x_train / 255.0)

# One-hot encoding
y_train = to_categorical(y_train, 10)
y_test = to_categorical(y_test, 10)

# 2. Definir modelo baseline
model = Sequential([
    Flatten(input_shape=(28,28)),
    Dense(128, activation='relu'),  # Sin regularización, sin dropout
    Dense(10, activation='softmax')
])

# 3. Compilar
model.compile(optimizer='adam',
              loss='categorical_crossentropy',
              metrics=['accuracy'])

# 4. Entrenar
history = model.fit(x_train, y_train,
                    validation_data=(x_test, y_test),
                    epochs=10, batch_size=32, verbose=2)

# 5. Evaluar
test_loss, test_acc = model.evaluate(x_test, y_test, verbose=0)
print(f"Exactitud en test (baseline): {test_acc:.4f}")

# 6. Graficar resultados
plot_history(history, title="Baseline (sin normalización ni dropout)")
```
